```python
top_channels = df.groupby('channel')['views'].sum().sort_values(ascending=False).head(10).reset_index()

fig = px.bar(top_channels, x='views', y='channel', orientation='h',
             title="🔥 Top 10 Channels by Total Views",
             color='views', color_continuous_scale='Viridis')
fig.update_layout(yaxis={'categoryorder': 'total ascending'})
fig.show()
```

```python
fig = px.histogram(df, x='published_hour', nbins=24,
                   title="⏰ Video Publish Time Distribution by Hour",
                   labels={'published_hour': 'Hour of Day'})
fig.update_layout(bargap=0.2)
fig.show()
```

```python
fig = px.histogram(df, x='published_day',
                   category_orders={"published_day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]},
                   title="📅 Trending Videos by Day of Week")
fig.show()
```

```python
fig = px.scatter(df, x='views', y='likes', size='comments', color='channel',
                 hover_data=['title'],
                 title='💙 Views vs Likes (Bubble Size = Comments)')
fig.update_layout(xaxis_type='log', yaxis_type='log')
fig.show()
```

```python
fig = px.scatter(df, x='like_ratio', y='comment_ratio', color='channel',
                 hover_name='title',
                 title='📊 Like Ratio vs Comment Ratio')
fig.show()
```
